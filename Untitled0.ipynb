```python
import pandas as pd

df_train = pd.read_csv('/content/customer_churn_dataset-training-master.csv')
print(df_train.head())
print(df_train.shape)
```

```python
print(df_train.dtypes)
print("\n")

print(df_train.isnull().sum())
print("\n")

print(df_train.describe())
```

```python
# Check for duplicates
print("Duplicates:", df_train.duplicated().sum())
print("\n")

# Remove rows with missing values
df_train = df_train.dropna()
print("Rows after removing missing:", df_train.shape)
print("\n")

# Remove duplicates
df_train = df_train.drop_duplicates()
print("Rows after removing duplicates:", df_train.shape)
print("\n")

# Check data again
print(df_train.isnull().sum())
```

```python
import sqlite3

# Create in-memory database
conn = sqlite3.connect(':memory:')

# Load dataframe into SQL table
df_train.to_sql('customers', conn, index=False, if_exists='replace')

# Query 1: Churn rate by subscription type
query1 = """
SELECT "Subscription Type",
       COUNT(*) as Total_Customers,
       SUM(Churn) as Churned_Customers,
       ROUND(SUM(Churn) * 100 / COUNT(*), 2) as Churn_Rate
FROM customers
GROUP BY "Subscription Type"
"""
print("Query 1: Churn by Subscription Type")
print(pd.read_sql(query1, conn))
print("\n")

# Query 2: Churn rate by contract length
query2 = """
SELECT "Contract Length",
       COUNT(*) as Total_Customers,
       SUM(Churn) as Churned_Customers,
       ROUND(SUM(Churn) * 100 / COUNT(*), 2) as Churn_Rate
FROM customers
GROUP BY "Contract Length"
"""
print("Query 2: Churn by Contract Length")
print(pd.read_sql(query2, conn))
```

```python
# Analysis 1: Compare churned vs non-churned customers
churned = df_train[df_train['Churn'] == 1]
not_churned = df_train[df_train['Churn'] == 0]

comparison = pd.DataFrame({
    'Churned_Avg': churned[['Age', 'Tenure', 'Support Calls', 'Payment Delay', 'Total Spend']].mean(),
    'Not_Churned_Avg': not_churned[['Age', 'Tenure', 'Support Calls', 'Payment Delay', 'Total Spend']].mean()
})
print("Analysis 1: Churned vs Not Churned Customers")
print(comparison)
print("\n")

# Analysis 2: Correlation with churn
print("Analysis 2: Correlation with Churn")
correlation = df_train[['Age', 'Tenure', 'Usage Frequency', 'Support Calls', 'Payment Delay', 'Total Spend', 'Churn']].corr()['Churn'].sort_values(ascending=False)
print(correlation)
print("\n")

# Analysis 3: High-risk customers
print("Analysis 3: Customers with Payment Delay > 20")
high_risk = df_train[df_train['Payment Delay'] > 20]['Churn'].sum()
total_high_risk = len(df_train[df_train['Payment Delay'] > 20])
print(f"Churned: {high_risk} out of {total_high_risk} = {(high_risk/total_high_risk*100):.2f}%")
```

```python
# Create summary dataframe
summary_data = {
    'Metric': ['Total Customers', 'Churned Customers', 'Churn Rate', 'Avg Support Calls (Churned)', 'Avg Support Calls (Not Churned)', 'Avg Payment Delay (Churned)', 'Avg Total Spend (Churned)'],
    'Value': [
        len(df_train),
        len(churned),
        f"{(len(churned)/len(df_train)*100):.2f}%",
        f"{churned['Support Calls'].mean():.2f}",
        f"{not_churned['Support Calls'].mean():.2f}",
        f"{churned['Payment Delay'].mean():.2f}",
        f"${churned['Total Spend'].mean():.2f}"
    ]
}
summary_df = pd.DataFrame(summary_data)

# Sheet 2: Churn by Subscription Type
churn_subscription = pd.read_sql(query1, conn)

# Sheet 3: Churn by Contract Length
churn_contract = pd.read_sql(query2, conn)

# Sheet 4: Churn by Age Groups
age_bins = [0, 25, 35, 45, 55, 65]
age_labels = ['18-25', '26-35', '36-45', '46-55', '56-65']
df_train['Age Group'] = pd.cut(df_train['Age'], bins=age_bins, labels=age_labels)
churn_age = df_train.groupby('Age Group').agg({
    'CustomerID': 'count',
    'Churn': 'sum'
}).rename(columns={'CustomerID': 'Total_Customers', 'Churn': 'Churned_Customers'})
churn_age['Churn_Rate'] = (churn_age['Churned_Customers'] / churn_age['Total_Customers'] * 100).round(2)
churn_age = churn_age.reset_index()

# Sheet 5: Top Risk Factors
risk_factors = pd.DataFrame({
    'Risk Factor': ['Support Calls > 8', 'Payment Delay > 20', 'Tenure < 10 months', 'Total Spend < 300'],
    'Churned %': [
        f"{(len(df_train[(df_train['Support Calls'] > 8) & (df_train['Churn'] == 1)]) / len(df_train[df_train['Support Calls'] > 8]) * 100):.2f}%",
        f"{(len(df_train[(df_train['Payment Delay'] > 20) & (df_train['Churn'] == 1)]) / len(df_train[df_train['Payment Delay'] > 20]) * 100):.2f}%",
        f"{(len(df_train[(df_train['Tenure'] < 10) & (df_train['Churn'] == 1)]) / len(df_train[df_train['Tenure'] < 10]) * 100):.2f}%",
        f"{(len(df_train[(df_train['Total Spend'] < 300) & (df_train['Churn'] == 1)]) / len(df_train[df_train['Total Spend'] < 300]) * 100):.2f}%"
    ]
})

# Export ALL sheets to Excel
with pd.ExcelWriter('Customer_Churn_Analysis.xlsx', engine='openpyxl') as writer:
    summary_df.to_excel(writer, sheet_name='Summary', index=False)
    churn_subscription.to_excel(writer, sheet_name='Churn by Subscription', index=False)
    churn_contract.to_excel(writer, sheet_name='Churn by Contract', index=False)
    churn_age.to_excel(writer, sheet_name='Churn by Age', index=False)
    risk_factors.to_excel(writer, sheet_name='Risk Factors', index=False)

print("Excel file created with 5 sheets: Customer_Churn_Analysis.xlsx")
print("\nSheets created:")
print("1. Summary")
print("2. Churn by Subscription")
print("3. Churn by Contract")
print("4. Churn by Age")
print("5. Risk Factors")
```

```python
# Export datasets for Tableau
df_train.to_csv('customer_churn_data.csv', index=False)
churn_subscription.to_csv('churn_by_subscription.csv', index=False)
churn_contract.to_csv('churn_by_contract.csv', index=False)
churn_age.to_csv('churn_by_age.csv', index=False)
risk_factors.to_csv('risk_factors.csv', index=False)

print("CSV files created:")
print("1. customer_churn_data.csv")
print("2. churn_by_subscription.csv")
print("3. churn_by_contract.csv")
print("4. churn_by_age.csv")
print("5. risk_factors.csv")
```
